# Numerical Computing with NumPy

NumPy is the foundation of scientific computing in Python. Its core innovation is the *ndarray*: a typed, multi-dimensional array that supports vectorised operations. Instead of looping over elements one by one, you write operations that apply to the whole array at once, and NumPy executes them in optimised C code. The speed difference over plain Python loops can be 10–100×.

**What's inside:** creating arrays, indexing and slicing, element-wise arithmetic, broadcasting, universal functions (ufuncs), aggregation along axes, reshaping, and linear algebra.

**Learn more:** [NumPy documentation](https://numpy.org/)

## 1. Creating Arrays

### 1.1 From a list

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5])
a
```

```python
# 2D array from nested lists
m = np.array([[1, 2, 3], [4, 5, 6]])
m
```

### 1.2 zeros / ones / full

```python
np.zeros((2, 4))
```

```python
np.ones((3, 2), dtype=int)
```

```python
np.full((2, 3), 7)
```

### 1.3 arange

Like Python's `range()` but returns an array.

```python
np.arange(0, 10, 2)   # start, stop (exclusive), step
```

### 1.4 linspace

Evenly spaced values between two endpoints, useful for plotting.

```python
np.linspace(0, 1, 5)   # 5 values from 0 to 1 inclusive
```

## 2. Array Attributes

```python
a = np.array([[1, 2, 3], [4, 5, 6]], dtype=float)

print(a.shape)   # (rows, columns)
print(a.ndim)    # number of dimensions
print(a.dtype)   # element type
print(a.size)    # total element count
```

## 3. Indexing and Slicing

### 3.1 1D indexing and slicing

```python
a = np.arange(10)
print(a[3])     # single element
print(a[-1])    # last element
print(a[2:7])   # slice
print(a[::2])   # every other element
```

### 3.2 2D indexing

`m[row, col]`: comma-separated indices instead of chained brackets.

```python
m = np.array([[10, 20, 30], [40, 50, 60], [70, 80, 90]])

print(m[1, 2])      # row 1, col 2
print(m[:, 1])      # entire column 1
print(m[0:2, 1:])   # sub-matrix
```

### 3.3 Boolean indexing

Pass a boolean array as the index; returns only elements where True.

```python
a = np.array([3, 7, 1, 9, 4])
a[a > 4]
```

```python
a[a % 2 == 0]   # even elements only
```

## 4. Arithmetic and Broadcasting

Operations are element-wise by default (no loops needed).

### 4.1 Element-wise arithmetic

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])

print(a + b)
print(a * b)
print(b / a)
```

### 4.2 Scalar broadcasting

A scalar is automatically applied to every element.

```python
a = np.array([1, 2, 3])
a * 10
```

### 4.3 Array broadcasting

Arrays with compatible shapes can be combined without explicit tiling.

```python
# 1D row array broadcast across each row of a 3x3 matrix
m = np.ones((3, 3), dtype=int)
row = np.array([1, 2, 3])
m + row
```

```python
# column array (shape 3x1) broadcast across each column
col = np.array([[10], [20], [30]])
m + col
```

## 5. Universal Functions (ufuncs)

NumPy provides vectorised math functions that operate on entire arrays at once.

```python
a = np.array([0, 1, 4, 9, 16])
np.sqrt(a)
```

```python
np.exp(np.array([0, 1, 2])).round(4)
```

```python
a = np.array([-3, -1, 0, 1, 3])
np.abs(a)
```

```python
a = np.array([1.2, 2.7, 3.5])
print(np.floor(a))
print(np.ceil(a))
print(np.round(a))
```

## 6. Aggregation Along Axes

`axis=0` collapses rows (result is one value per column); `axis=1` collapses columns.

```python
a = np.array([[1, 2, 3], [4, 5, 6]])

print(a.sum())
print(a.mean())
print(a.std().round(4))
print(a.min(), a.max())
```

```python
# axis=0: collapse rows (result is one value per column)
a.sum(axis=0)
```

```python
# axis=1: collapse columns (result is one value per row)
a.sum(axis=1)
```

```python
a.mean(axis=1)
```

## 7. Reshaping

### 7.1 reshape

Change the shape without changing the data. Use -1 to let NumPy infer a dimension.

```python
a = np.arange(12)
a.reshape(3, 4)
```

```python
a.reshape(4, -1)   # -1 means 'infer this dimension'
```

### 7.2 flatten and ravel

```python
m = np.array([[1, 2], [3, 4]])

print(m.flatten())   # always returns a copy
print(m.ravel())     # returns a view when possible (more efficient)
```

### 7.3 Transpose

```python
m = np.array([[1, 2, 3], [4, 5, 6]])
m.T
```

## 8. Linear Algebra

### 8.1 Dot product (1D vectors)

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
np.dot(a, b)   # 1*4 + 2*5 + 3*6 = 32
```

### 8.2 Matrix multiplication

The `@` operator performs matrix multiplication.

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
A @ B
```

### 8.3 Determinant and inverse

```python
A = np.array([[1, 2], [3, 4]], dtype=float)

print(np.linalg.det(A))
print(np.linalg.inv(A))
```
