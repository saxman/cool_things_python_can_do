# Python Dictionaries

The Python dict is one of the most powerful and frequently used data structures in the language. Beyond basic key-value storage, dicts support comprehensions, elegant merging syntax, and can even replace long `if/elif` chains. Lookups are O(1) on average regardless of dictionary size.

**What's inside:** creating dicts, comprehensions, merging, safe access with defaults, iterating, sorting by value, nested dicts, and using dicts as dispatch tables.

**Learn more:** [Python dict](https://docs.python.org/3/library/stdtypes.html#dict)

## 1. Creating Dictionaries

### 1.1 Literal and dict()

```python
# literal syntax
scores = {'Alice': 88, 'Bob': 74, 'Carol': 95}
scores
```

```python
# dict() with keyword arguments (keys must be valid identifiers)
dict(Alice=88, Bob=74, Carol=95)
```

### 1.2 From two lists with zip

```python
names  = ['Alice', 'Bob', 'Carol']
values = [88, 74, 95]

dict(zip(names, values))
```

### 1.3 Dict comprehensions

The dict equivalent of list comprehensions: build a dict from any iterable in one expression.

```python
# squares of numbers 1-5
{n: n**2 for n in range(1, 6)}
```

```python
# filter while building
{name: score for name, score in scores.items() if score >= 80}
```

```python
# transform values
{name: score / 100 for name, score in scores.items()}
```

### 1.4 fromkeys: initialise keys with a default value

```python
# useful for initialising counters or flags
dict.fromkeys(['a', 'b', 'c'], 0)
```

## 2. Accessing Data

### 2.1 get(): safe access with a default

`d[key]` raises `KeyError` if the key is absent; `d.get(key, default)` returns the default instead.

```python
print(scores.get('Alice'))       # 88
print(scores.get('Dave'))        # None
print(scores.get('Dave', 0))     # 0 (explicit default)
```

### 2.2 setdefault(): insert if absent, then return

Inserts a key with a default value if it doesn't exist, then returns the value. Useful for accumulating into a dict.

```python
# group names by first letter
groups = {}
for name in ['Alice', 'Bob', 'Anna', 'Brian', 'Carol']:
    groups.setdefault(name[0], []).append(name)

groups
```

### 2.3 Membership testing

```python
print('Alice' in scores)    # True
print('Dave' in scores)     # False
print('Dave' not in scores) # True
```

## 3. Modifying Dictionaries

### 3.1 Adding and updating

```python
d = {'a': 1, 'b': 2}
d['c'] = 3       # add
d['a'] = 10      # update
d
```

### 3.2 Merging dictionaries

Python 3.9+ introduced the `|` operator for merging; the right operand wins on key conflicts.

```python
defaults = {'timeout': 30, 'retries': 3, 'verbose': False}
overrides = {'timeout': 10, 'verbose': True}

# | creates a new merged dict
config = defaults | overrides
config
```

```python
# |= merges in place
defaults |= overrides
defaults
```

### 3.3 Removing items

```python
d = {'a': 1, 'b': 2, 'c': 3}

val = d.pop('b')          # remove and return value
print(val, d)

d.pop('z', None)          # safe pop (no KeyError if missing)
print(d)
```

```python
d = {'a': 1, 'b': 2, 'c': 3}
key, val = d.popitem()    # removes and returns the last inserted item
print(key, val)
print(d)
```

## 4. Iterating

### 4.1 keys(), values(), items()

```python
for name, score in scores.items():
    print(f'{name}: {score}')
```

```python
# iterating a dict directly yields keys
list(scores)           # equivalent to list(scores.keys())
```

### 4.2 Sorting by value

```python
# sort by score, highest first
sorted(scores.items(), key=lambda item: item[1], reverse=True)
```

```python
# rebuild as a sorted dict
dict(sorted(scores.items(), key=lambda item: item[1], reverse=True))
```

### 4.3 min and max by value

```python
print(max(scores, key=scores.get))   # name with highest score
print(min(scores, key=scores.get))   # name with lowest score
```

## 5. Nested Dictionaries

```python
students = {
    'Alice': {'age': 24, 'grade': 'A', 'city': 'Austin'},
    'Bob':   {'age': 31, 'grade': 'C', 'city': 'Boston'},
}

students['Alice']['grade']
```

```python
# comprehension over nested dict
{name: info['grade'] for name, info in students.items()}
```

```python
# safe deep access with chained get()
students.get('Dave', {}).get('grade', 'unknown')
```

## 6. Dict as a Dispatch Table

A dict of functions can replace a long `if/elif` chain: more readable and O(1) lookup.

```python
# verbose if/elif approach
def handle_v1(event):
    if event == 'start':   return 'starting...'
    elif event == 'stop':  return 'stopping...'
    elif event == 'pause': return 'pausing...'
    else:                  return 'unknown event'
```

```python
# dispatch table approach
handlers = {
    'start': lambda: 'starting...',
    'stop':  lambda: 'stopping...',
    'pause': lambda: 'pausing...',
}

def handle(event):
    return handlers.get(event, lambda: 'unknown event')()

print(handle('start'))
print(handle('pause'))
print(handle('reset'))
```
